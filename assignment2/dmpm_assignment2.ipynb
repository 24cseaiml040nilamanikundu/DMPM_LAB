```python
#Q1
num = list(map(int, input("Enter the list of numbers: ").split(",")))

total = 0
count = 0

for i in num:
    total += i
    count += 1

mean = total / count
print("Mean =", mean)
```

```python
#Q2
num = list(map(int, input("Enter the list of numbers: ").split(",")))
mean = sum(num) / len(num)
print("The mean of the data points is:", mean)
```

```python
#Q3
num = list(map(int, input("Enter the list of numbers: ").split(",")))
num.sort()
n = len(num)

if n % 2 == 0:
    median = (num[n//2] + num[n//2-1]) / 2
else:
    median = num[n+1// 2]
print(f"The median is: {median}")
```

```python
#Q4
def find_median(arr):
    arr.sort()
    n = len(arr)

    if n % 2 == 1:
        return arr[n // 2]
    else:
        return (arr[n // 2 - 1] + arr[n // 2]) / 2

numbers = list(map(int, input("Enter numbers separated by spaces: ").split(",")))

print("Median =", find_median(numbers))
```

```python

#Q5
l = [int(x) for x in input("Enter the list of number :").split(",")]
dict = {}
for x in l:
    if x in dict:
        dict[x] += 1
    else:
        dict[x] = 1
mode_without = max(dict, key=dict.get)
print(mode_without)
```

```python
#Q7
l = [int(x) for x in input("Enter the list of number :").split(",")]
mean = sum(l)/len(l)
sum_squared_diff = sum((x - mean) ** 2 for x in l)
variance = sum_squared_diff / len(l)
print(f"Varience without using function : {variance:.4f}")
```

```python
#Q8
import numpy as np
l = [int(x) for x in input("Enter the list of number :").split(",")]
l = np.array(l)
variance = np.var(l)
print(f"Varience with using function : {variance:.4f}")
```

```python
#Q9
l = [int(x) for x in input("Enter the list of number :").split(",")]
mean = sum(l)/len(l)
sum_squared_diff = sum((x - mean) ** 2 for x in l)
sd = (sum_squared_diff / len(l))**0.5
print(f"Standard Deviation without function : {sd:.4f}")

l2 = np.array(l)
sd = np.std(l2)
print(f"Standard Deviation with using function : {sd:.4f}")
```
