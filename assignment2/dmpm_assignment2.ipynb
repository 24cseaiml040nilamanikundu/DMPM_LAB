```python
num = [32, 34, 35, 35, 36, 37, 38, 39, 39, 47]

n = len(num)

if n % 2 == 0:
    mid1 = n // 2
    mid2 = n // 2 - 1
    median = (num[mid1] + num[mid2]) / 2
else:
    median = num[n+1// 2]

print(f"Sorted list: {num}")
print(f"The median is: {median}")
```

```python
def find_median(arr):
    arr.sort()
    n = len(arr)

    if n % 2 == 1:
        return arr[n // 2]
    else:
        return (arr[n // 2 - 1] + arr[n // 2]) / 2

numbers = [32, 34, 35, 35, 36, 37, 38, 39, 39, 47]
print("Median =", find_median(numbers))
```

```python
def find_mode(a):
    max_count = 0
    mode = a[0]

    for i in range(len(a)):
        count = 0
        for j in range(len(a)):
            if a[i] == a[j]:
                count += 1

        if count > max_count:
            max_count = count
            mode = a[i]

    return mode

n = int(input("Enter number of elements: "))

a = []
for i in range(n):
    a.append(int(input("Enter data:")))

print("Mode =", find_mode(a))
```
