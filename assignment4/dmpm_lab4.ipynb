```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =========================================================
# Load the Dataset
# =========================================================

url = "https://raw.githubusercontent.com/24cseaiml040nilamanikundu/DMPM_LAB/refs/heads/main/assignment4/Cancer_Data.csv"

data = pd.read_csv(url)

# Remove unnecessary columns if they exist
data = data.drop(columns=["Unnamed: 32"], errors="ignore")



# =========================================================
# 1. Box Plot: Radius Distribution by Diagnosis
# =========================================================

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=data,
    x="diagnosis",
    y="radius_mean",
    hue="diagnosis",
    legend=False
)

plt.title("1. Box Plot: Radius Mean by Diagnosis")
plt.xlabel("Diagnosis (B = Benign, M = Malignant)")
plt.ylabel("Radius Mean")

plt.show()


# =========================================================
# 2. Scatter Plot: Relationship between Radius and Texture
# =========================================================

plt.figure(figsize=(7, 5))

sns.scatterplot(
    data=data,
    x="radius_mean",
    y="texture_mean",
    hue="diagnosis"
)

plt.title("2. Scatter Plot: Radius Mean vs Texture Mean")
plt.xlabel("Radius Mean")
plt.ylabel("Texture Mean")
plt.legend(title="Diagnosis")

plt.show()


# =========================================================
# 3. Histogram: Distribution of Area Mean
# =========================================================

plt.figure(figsize=(7, 5))

plt.hist(
    data["area_mean"],
    bins=20,
    color="skyblue",
    edgecolor="black"
)

plt.title("3. Histogram: Distribution of Area Mean")
plt.xlabel("Area Mean")
plt.ylabel("Count")

plt.show()


# =========================================================
# 4. Line Chart: Radius Mean for First 30 Samples
# =========================================================

first_30 = data["radius_mean"].iloc[:30]

plt.figure(figsize=(8, 5))

plt.plot(
    first_30.index,
    first_30.values,
    marker="o",
    color="green",
    label="Radius Mean"
)

plt.title("4. Line Chart: Radius Mean for First 30 Samples")
plt.xlabel("Sample Index")
plt.ylabel("Radius Mean")

plt.grid(True)
plt.legend()

plt.show()


# =========================================================
# 5. Pie Chart: Diagnosis Distribution
# =========================================================

diagnosis_count = data["diagnosis"].value_counts()

plt.figure(figsize=(6, 6))

plt.pie(
    diagnosis_count,
    labels=diagnosis_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("5. Pie Chart: Diagnosis Distribution")

plt.show()


# =========================================================
# 6. Bar Plot: Average Area Mean by Diagnosis
# =========================================================

area_average = data.groupby("diagnosis")["area_mean"].mean()

plt.figure(figsize=(7, 5))

sns.barplot(
    x=area_average.index,
    y=area_average.values,
    hue=area_average.index,
    legend=False
)

plt.title("6. Bar Plot: Average Area Mean by Diagnosis")
plt.xlabel("Diagnosis")
plt.ylabel("Average Area Mean")

plt.show()
```
