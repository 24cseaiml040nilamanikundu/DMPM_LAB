```python
#implement z-score normalization,min-max normalization, and decimal scalling normalization.
import numpy as np
arr=np.array([10,20,30,40,50])

#z-score normalization
print("z-score normalization")
mean=np.mean(arr)
std=np.std(arr)
z_score=(arr-mean)/std
print(z_score)

#min-max normalization
print("\nmin-max normalization")
min=np.min(arr)
max=np.max(arr)
min_max_norm=(arr-min)/(max-min)
print(min_max_norm)

#decimal scalling normalization
print("\ndecimal scalling normalization")
j=len(str(np.max(arr)))
decimal_scalling=arr/(10**j)
print(decimal_scalling)
```

```python
#implement the decimal scalling normalization on a dataset from kaggel.
import pandas as pd

df=pd.read_csv("https://raw.githubusercontent.com/24cseaiml040nilamanikundu/DMPM_LAB/refs/heads/main/Assignment5/results.csv")

j=len(str(np.max(df['English'])))
decimal_scalling=df['English']/(10**j)
print(decimal_scalling)
```

```python
#implement the z-score normalization on any dataset from kaggel and display the mean,varience and standard deviation.
df=pd.read_csv("https://raw.githubusercontent.com/24cseaiml040nilamanikundu/DMPM_LAB/refs/heads/main/Assignment5/results.csv")

mean=np.mean(df['English'])
var=np.var(df['English'])
std=np.std(df['English'])
z_score=(df['English']-mean)/std
print("mean:",mean)
print("variance:",var)
print("standard deviation:",std)
print("\nZ-Score Normalization:")
print(z_score)
```
